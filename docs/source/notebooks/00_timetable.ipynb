# Kjøring av språkmodeller på egen maskin med Ollama

Dette er et introduksjonskurs hvor du lærer å kjøre Språkmodellene lokalt på egen maskin. Språkmodeller som ChatGPT, er trenet til å generere ordsekvenser basert på kontekst, som når de fullfører setningen. For eksempel, hvis vi begynner med “kyllingen krysset veien”, kan en mulig fullføring være “for å finne mat” eller “komme over på den andre siden”. For å lære mer om hvordan språkmodeller trenes, anbefaler vi å se youtube videoene [Intro to Large Language Models](https://www.youtube.com/watch?v=zjkBMFhNj_g) og [Deep Dive into LLMs like ChatGPT](https://www.youtube.com/watch?v=7xTGNNLPyMI) av Andrej Karpathy.

I dette kurset bruker vi Python programmering til å kjøre LLMer. Dette er grunnen til at vi forutsetter grunnleggende programmeringskunnskaper. Hvis programmering er nytt for deg, anbefaler vi at du øver litt på forhånd. Carpentry leksjonen [Plotting and Programming in Python](https://swcarpentry.github.io/python-novice-gapminder/), bør tas som selvstudiom eller kurs med påmelding,

Kurset er sammensatte av flere kapitler. I [Oppstart](01_easy_login.ipynb), logger vi inn på Fox.

I [Installering](02_downloading_packages.ipynb) installerer vi programvaren som vi trenger.

I [Instruks til Store Språkmodeller (Chatboter)](03_chatbot.ipynb) lærer vi om opplasting og grunnleggende spørring av LLMer.

I [Oppsummeringer](04_summarization.ipynb) bruker vi LLMer til å lage korte oppsummeringer av dokumenter.

I [Retrieval-Augmented Generation (RAG)](05_rag.ipynb), bygger vi et program som har hentingsforsterket tekstgenerering med noen dokumenter som vi bruker som eksempler.
