## Forberedelser


```{admonition} Oppgave 2.1: Dokumentmappe
Du trenger en mappe til å lagre dokumentene dine på maskinen din. Når du åpner JupyterLab, vil du få filmenyen til venstre. Den peker til ditt hjemmeområde/ home directory som du kan finne igjen i terminalen. Det anbefales at man har orden i prosjektene siden ved at man starter på stien cd ~, og deretter lager en prosjektmappe i snake-case. Her kan du lagre filene dine. Neste gang du skal jobbe på prosjektet ditt setter du setwd, og stien til prosjektet som du laget.

Hvis du ikke allerede har en mappe som kalles “documents” inni prosjektet ditt, lag en. Du kan bruke terminal med bash kommandoer, eller jupyter lab.

I jupyterlab: For å lage en ny mappe, klikk den grå , “New Folder” knappen i toppmenyen. Nå dukker det en ny mappe opp på listen, med forslag til navn “Untitled Folder”. Skriv “documents” istedenfor “Untitled Folder” og trykk enter på tastaturet for å lagre.
```

```{admonition} Oppgave 2.2: Laste opp filer til mappen
Du kan laste opp dokumenter som du ønsker å jobbe med. Lukk alle programer, og sjekk om du finner mappen igjen.
```
